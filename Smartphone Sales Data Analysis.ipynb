SMARTPHONE SALES DATASET

This dataset contains information about different mobile phone brands and models available in the market. It also provides customer ratings, which help in understanding customer satisfaction and product preferences. This dataset provides details about mobile phone specifications, customer ratings, selling prices, original prices and discounts.
The dataset includes information about popular mobile phone brands such as Apple, Samsung, Nokia, Realme, Infinix and other manufacturers. It contains product-related details such as model names, available colours, RAM (Memory), storage capacity and camera availability. It includes both original prices and selling prices, allowing comparisons of product prices and the calculation of discounts and discount percentages. These details make it useful for analysing pricing strategies, brand comparisons and discount trends in the mobile phone market.

COLOUM WISE DESCRIPTION

1. Brand : Thi coloum indicate the different brands available in the market

2. Model : This coloum is to identify and understand different models or varients of brands phone available in the current market.

3. Colors : This coloumn indicates the colour of each model or varient phones

4. Memory : Thi coloum represents the memory or RAM capacity of each phones in the dataset

5. Storage : This coloum represents the storage availability in each phones in the dataset

6. Camera : This coloum indicates that the specified hone has camera or not

7. Rating : Thi coloum represents the rating of each phone in the current period out of 5 

8. Selling Price : This coloum indicates at which price does the phone is selling

9. Orginal Price : This coloum indicates the actual price of the phone 

10. Mobile : This coloum represent the different brand phone available in the market

11. Discount : Thi coloum represent at how much discount rate the phone is selled [ orginal price - selling price ]

12. Dicount Percentage : This coloum represent at how much percantage the discount is applicable for the mobile phones [ discount/actual price*100]
